## 2. Εντοπισμός ελλείψεων
Η `isna()` εντοπίζει αναγνωρισμένες ελλείψεις, όπως `pd.NA`. **Δεν θεωρεί το `""`, το `" "` ή το `"-"` αυτομάτως ελλείψεις.** Στην παρούσα ρητή εισαγωγή θα δούμε αρχικά μηδενικές αναγνωρισμένες ελλείψεις, παρότι λείπει πληροφορία.

Στη συνέχεια εφαρμόζουμε τη σύμβαση του λεξικού: κενό έτος και `-` στον τόπο δηλώνουν άγνωστο. Η `.replace()` παρακάτω αντικαθιστά συγκεκριμένες τιμές σε συγκεκριμένα πεδία. Η `.str.strip()` αφαιρεί κενά από τα άκρα του κειμένου.

## 3. Ακριβή αντίγραφα και συγκρούσεις
Η `duplicated()` συγκρίνει όλες τις στήλες και, από προεπιλογή, σημαίνει ως `True` κάθε επιπλέον αντίγραφο μετά το πρώτο. Με `keep=False` επισημαίνει **όλες** τις γραμμές που ανήκουν σε ομάδα αντιγράφων. Ο δείκτης γραμμής δεν συμμετέχει.

Η `drop_duplicates()` κρατά το πρώτο ακριβές αντίγραφο. Εδώ η περιγραφή της εξαγωγής βεβαιώνει ότι πρόκειται για πλεονάζουσα γραμμή, και η μονάδα απαιτεί ένα τεκμήριο ανά γραμμή.

```python
print("Πλεονάζοντα ακριβή αντίγραφα:", work.duplicated().sum())
display(work.loc[work.duplicated(keep=False), ["object_id", "title", "place"]])

exact_removed = work.loc[work.duplicated()].copy()
no_exact = work.drop_duplicates().copy()
print("Γραμμές μετά την αφαίρεση ακριβούς αντιγράφου:", len(no_exact))
```

## 4. Κείμενο και κατηγορίες
Θα δημιουργήσουμε συγκρίσιμη μορφή κειμένου με:

- `.str.normalize("NFC")`: ενοποίηση κανονικά ισοδύναμων αναπαραστάσεων Unicode, π.χ. γράμματος με συνδυαστικό τόνο. Δεν αφαιρεί τους τόνους και δεν διορθώνει λέξεις.
- `.str.strip()`: αφαίρεση κενών στα άκρα.
- `.str.replace(r"\s+", " ", regex=True)`: έτοιμος κανόνας που αντικαθιστά κάθε ακολουθία κενών χαρακτήρων με ένα κενό. Το `r"\s+"` είναι μοτίβο κανονικής έκφρασης .
- `.str.casefold()`: μορφή κατάλληλη για σύγκριση χωρίς διάκριση πεζών/κεφαλαίων.

```python
title_before = work["title"].copy()
work["title"] = work["title"].str.normalize("NFC").str.strip()
work["title"] = work["title"].str.replace(r"\s+", " ", regex=True)

place_key = work["place"].str.normalize("NFC").str.strip().str.casefold()
place_labels = {"αθήνα": "Αθήνα", "αθηνα": "Αθήνα", "σύροσ": "Σύρος", "πάτρα": "Πάτρα"}
work["place"] = place_key.replace(place_labels)

display(work[["object_id", "title", "place"]].head(6))
```

**Γιατί το κλειδί είναι `"σύροσ"`;** Η `casefold()` ενοποιεί και το τελικό `ς` με το `σ` για σύγκριση. Η τελική ετικέτα `"Σύρος"` προκύπτει από τη ρητή αντιστοίχιση. Δεν αλλάζουμε την ιστορική ονομασία. Τα άτονα κλειδιά, όπως αθηνα και φωτογραφια, αντιστοιχίζονται ρητά επειδή εμφανίζονται στις κεφαλαίες γραφές του συγκεκριμένου αρχείου. Δεν αφαιρούμε αυτόματα τόνους από κάθε λέξη.

Για τα είδη, εφαρμόζουμε επίσης ρητό λεξικό ετικετών. Η `replace()` αφήνει άγνωστες τιμές όπως είναι. Γι’ αυτό ακολουθεί έλεγχος `.isin(...)` που εξετάζει αν κάθε τιμή ανήκει στην επιτρεπτή λίστα. Δεν θέλουμε άγνωστες κατηγορίες να γίνουν σιωπηρά ελλείψεις.

```python
type_key = work["object_type"].str.normalize("NFC").str.strip().str.casefold()
type_labels = {
    "φωτογραφία": "Φωτογραφία", "φωτογραφια": "Φωτογραφία",
    "επιστολή": "Επιστολή", "επιστολη": "Επιστολή",
    "χάρτησ": "Χάρτης",
    "αφίσα": "Αφίσα",
    "κατάστιχο": "Κατάστιχο", "καταστιχο": "Κατάστιχο"
}
work["object_type"] = type_key.replace(type_labels)
allowed_types = ["Φωτογραφία", "Επιστολή", "Χάρτης", "Αφίσα", "Κατάστιχο"]
unknown_type = ~work["object_type"].isin(allowed_types)
print("Τιμές εκτός επιτρεπτών κατηγοριών:", unknown_type.sum())
display(work[["object_id", "object_type"]])
```

## 5. Αριθμητικοί τύποι
Η `pd.to_numeric(..., errors="coerce")` διαβάζει ό,τι μπορεί ως αριθμό. Όταν μια μη κενή τιμή δεν μετατρέπεται, επιστρέφει έλλειψη. Αυτό **δεν είναι διόρθωση του περιεχομένου**.

Θα κρατήσουμε το `creation_year_text` ως τη δηλωμένη γραφή μετά την αναγνώριση κενών. Η αρχική, απολύτως αμετάβλητη γραφή παραμένει επίσης στο `raw`.

Η `notna()` είναι το αντίθετο της `isna()`. Το `&` συνδυάζει δύο συνθήκες ανά γραμμή. Το `failed_year` σημαίνει: «είχα μη κενό κείμενο, αλλά δεν πήρα αριθμό».

```python
work["creation_year_text"] = work["creation_year"].copy()
year_first = pd.to_numeric(work["creation_year_text"], errors="coerce")
failed_year = work["creation_year_text"].notna() & year_first.isna()

display(work.loc[failed_year, ["object_id", "creation_year_text"]])
print("Αποτυχημένες μετατροπές μη κενών τιμών:", failed_year.sum())
```

Οι δύο αποτυχίες έχουν διαφορετικό νόημα:
- Το `19O2` του OBJ006 περιέχει γράμμα O. Χρειάζεται έλεγχο πριν γίνει διόρθωση.
- Το `περ. 1900` του OBJ009 είναι προσεγγιστική χρονολόγηση. Δεν είναι απλώς «κακογραμμένος ακέραιος».

Ο κανόνας εγκυρότητας για τα γνωστά ακριβή έτη είναι: ακέραιος από 1870 έως 1930. Το `% 1` ελέγχει αν υπάρχει κλασματικό υπόλοιπο. Το `|` συνδυάζει συνθήκες με «ή». Οι συγκρίσεις γράφονται σε παρενθέσεις.

```python
invalid_year_first = year_first.notna() & (
    (year_first < 1870) | (year_first > 1930) | (year_first % 1 != 0)
)
display(work.loc[invalid_year_first, ["object_id", "creation_year_text"]])
```

Με `work.loc[work["object_id"] == "OBJ006", "creation_year"] = ...` επιλέγουμε συγκεκριμένη γραμμή και πεδίο. Η αντιστοίχιση γίνεται με το σταθερό αναγνωριστικό, όχι με θέση γραμμής που μπορεί να έχει αλλάξει.

```python
year_input = work["creation_year_text"].copy()
year_input.loc[work["object_id"] == "OBJ006"] = "1902"
year_input.loc[work["object_id"] == "OBJ008"] = "1898"

# Ελέγχουμε αριθμητικό περιεχόμενο πριν από τη μετατροπή σε Int64.
year_numeric = pd.to_numeric(year_input, errors="coerce")
invalid_year = year_numeric.notna() & (
    (year_numeric < 1870) | (year_numeric > 1930) | (year_numeric % 1 != 0)
)
print("Γνωστά έτη εκτός κανόνα μετά τον έλεγχο:", invalid_year.sum())
year_numeric = year_numeric.mask(invalid_year, pd.NA)
work["creation_year"] = year_numeric.astype("Int64")
work["date_is_approx"] = work["creation_year_text"].str.startswith("περ.", na=False)

audit_events = [
    ["OBJ006", "creation_year", "19O2", "1902", "Διόρθωση", "Έλεγχος ακριβούς έτους", "REF-A"],
    ["OBJ008", "creation_year", "2098", "1898", "Διόρθωση", "Έλεγχος ακριβούς έτους", "REF-B"]
]
display(work[["object_id", "creation_year_text", "creation_year", "date_is_approx"]])
```

Η `.mask(συνθήκη, pd.NA)` αντικαθιστά με έλλειψη τις θέσεις όπου η συνθήκη είναι True. Εδώ ο τελικός έλεγχος δεν βρήκε νέα μη έγκυρα γνωστά έτη.

Το `Int64` με κεφαλαίο I είναι ακέραιος τύπος της pandas που δέχεται και ελλείψεις. Πριν τον χρησιμοποιήσουμε, ελέγξαμε και τις κλασματικές τιμές. Η `.str.startswith("περ.", na=False)` σημειώνει το συγκεκριμένο μοτίβο προσεγγιστικής χρονολόγησης της άσκησης, χωρίς να θεωρεί ότι αναγνωρίζει κάθε ιστορική έκφραση χρόνου.

### Αρνητικά, κλασματικά και ασυνήθιστα πλήθη
Πρώτα μετατρέπουμε σε αριθμούς και ελέγχουμε αποτυχίες μετατροπής. Έπειτα ελέγχουμε τον κανόνα: μη αρνητικός ακέραιος. Ένα μεγάλο πλήθος δεν παραβιάζει αυτόν τον κανόνα.

Για το −2 δεν υπάρχει τεκμηριωμένη σωστή τιμή. Θα γίνει `pd.NA` στο κύριο αριθμητικό πεδίο, ενώ η αρχική τιμή θα διατηρηθεί στο `image_count_raw` και στο `raw`. Δεν το κάνουμε +2, ούτε 0.

```python
work["image_count_raw"] = work["image_count"].copy()
images = pd.to_numeric(work["image_count_raw"], errors="coerce")
failed_images = work["image_count_raw"].notna() & images.isna()
invalid_images = images.notna() & ((images < 0) | (images % 1 != 0))
print("Αποτυχίες μετατροπής πλήθους:", failed_images.sum())
display(work.loc[invalid_images, ["object_id", "image_count_raw"]])

work["image_count"] = images.mask(invalid_images, pd.NA).astype("Int64")
audit_events.extend([
    ["OBJ005", "image_count", "-2", "<NA>", "Μη έγκυρη τιμή", "Μη αρνητικό ακέραιο πλήθος· σωστή τιμή άγνωστη", "Κανόνας πεδίου"],
    ["OBJ010", "image_count", "300", "300", "Διατήρηση μετά από έλεγχο", "Πολυσέλιδο κατάστιχο με 300 εικόνες", "REF-C"]
])
display(work.loc[work["object_id"].isin(["OBJ005", "OBJ010"]), ["object_id", "object_type", "image_count_raw", "image_count"]])
```

## 7. Κατάλληλες χρήσεις των dropna και fillna
Ο κύριος πίνακας περιλαμβάνει τεκμήρια με άγνωστο ακριβές έτος. Για συγκεκριμένη χρονική ανάλυση μπορούμε να δημιουργήσουμε ξεχωριστό υποσύνολο:

`work.dropna(subset=["creation_year"])`

Χωρίς `subset`, η προεπιλεγμένη `dropna()` αφαιρεί κάθε γραμμή που έχει έλλειψη **σε οποιαδήποτε στήλη**. Στον πίνακά μας θα απέκλειε και χρήσιμα τεκμήρια που έχουν άγνωστο τόπο ή πλήθος εικόνων.

Η `fillna()` μπορεί να χρησιμοποιηθεί για **ετικέτα προβολής**: νέα στήλη `place_display` που γράφει «Μη καταγεγραμμένος». Το αρχικό `place` παραμένει ελλιπές και η γνώση μας για τον τόπο δεν αυξάνεται.

```python
print("Κύριος πίνακας:", len(work))
print("dropna() σε όλες τις στήλες:", len(work.dropna()))
print("dropna() μόνο στο ακριβές έτος:", len(work.dropna(subset=["creation_year"])))

work["place_display"] = work["place"].fillna("Μη καταγεγραμμένος")
display(work.loc[work["place"].isna(), ["object_id", "place", "place_display"]])
```

## 8. Ημερολόγιο και σύγκριση
Το ημερολόγιο πρέπει να περιλαμβάνει και τις **αποφάσεις διατήρησης ή εκκρεμότητας**, όχι μόνο αλλαγές τιμών. Οι μαζικοί κανόνες κειμένου θα καταγραφούν επίσης με το πραγματικό πλήθος πεδίων που άλλαξαν.

Το επόμενο κελί είναι έτοιμο. Δημιουργεί πίνακα αλλαγών και υπολογίζει τις αλλαγές κειμένου συγκρίνοντας κάθε τελικό ID με την πρώτη αρχική εγγραφή του. Η επιλογή πρώτης γραμμής εδώ χρησιμοποιείται **μόνο για σύγκριση με το raw**: το ανεπίλυτο OBJ007 έχει ήδη απομονωθεί και δεν συμμετέχει στον κύριο πίνακα.

```python
# Ξεκινά από αντίγραφο της λίστας ώστε η επανεκτέλεση αυτού του κελιού
# να μην προσθέτει επανειλημμένα τις ίδιες εγγραφές.
audit_rows = list(audit_events)
audit_rows.extend([
    ["OBJ003", "ολόκληρη γραμμή", "2 ακριβή αντίγραφα", "1 αντίγραφο", "Αφαίρεση 1 πλεονάζουσας γραμμής", "Επανάληψη εξαγωγής· μία μονάδα ανά τεκμήριο", "Κανόνας συλλογής"],
    ["OBJ007", "place", "Αθήνα / Πάτρα", "Εκκρεμότητα", "Προσωρινή εξαίρεση 2 γραμμών", "Ανεπίλυτη σύγκρουση", "conflicts.csv"],
    ["OBJ002", "creation_year", "κενό", "<NA>", "Διατήρηση άγνωστου έτους", "Δεν υπάρχει τεκμηριωμένη χρονολογία", "Λεξικό πεδίου"],
    ["OBJ009", "creation_year", "περ. 1900", "<NA> + αρχικό κείμενο", "Διατήρηση αβεβαιότητας", "Δεν έχει τεκμηριωθεί ακριβές έτος", "creation_year_text"],
    ["OBJ012", "place", "-", "<NA>", "Ενοποίηση έλλειψης", "Δηλωμένο σύμβολο άγνωστου τόπου", "Λεξικό πεδίου"]
])
raw_lookup = raw.drop_duplicates(subset=["object_id"]).set_index("object_id")
text_changes = []
for _, row in work.iterrows():
    for field in ["title", "object_type", "place"]:
        before = raw_lookup.loc[row["object_id"], field]
        after = row[field]
        if pd.notna(after) and before != after:
            text_changes.append([row["object_id"], field, before, after,
                "Κανονικοποίηση", "NFC/κενά/ελεγχόμενη ετικέτα ανά πεδίο", "TEXT-v1"])
audit_rows.extend(text_changes)
audit_log = pd.DataFrame(audit_rows, columns=[
    "object_id", "field", "before", "after", "action", "reason", "evidence"
])
print("Καταγεγραμμένες αλλαγές κειμενικών πεδίων:", len(text_changes))
display(audit_log)
```

### Σύγκριση με σαφείς ορισμούς
Θα συγκρίνουμε τις ίδιες μετρήσεις πριν και μετά. Επειδή το πλήθος γραμμών αλλάζει, κάθε αποτέλεσμα αναφέρεται ρητά στο αντίστοιχο μέγεθος δείγματος.

- **Κενό ακριβές έτος μετά την ερμηνεία:** έλλειψη στο αριθμητικό έτος, συμπεριλαμβανομένων αρχικών μη αριθμητικών τιμών.
- **Αριθμητικό έτος εκτός κανόνα:** γνωστός αριθμός εκτός 1870–1930 ή μη ακέραιος.
- **Μη έγκυρο γνωστό πλήθος εικόνων:** αρνητικός ή κλασματικός αριθμός.
- Η στήλη `place_display` δεν χρησιμοποιείται για να μετρήσουμε πληρότητα: είναι απλώς ετικέτα προβολής.

Το παρακάτω έτοιμο κελί δεν αλλάζει δεδομένα. Η πρόσθετη υποστήριξη από τις κάρτες αναφοράς φαίνεται στο ημερολόγιο, όχι μόνο στους αριθμούς.

```python
raw_year = pd.to_numeric(raw["creation_year"].replace("", pd.NA), errors="coerce")
raw_images = pd.to_numeric(raw["image_count"], errors="coerce")
raw_place = raw["place"].str.strip().replace({"": pd.NA, "-": pd.NA})

summary_rows = [
    ["Γραμμές", len(raw), len(work)],
    ["Διαφορετικά object_id", raw["object_id"].nunique(), work["object_id"].nunique()],
    ["Πλεονάζουσες γραμμές βάσει ID", raw.duplicated(subset=["object_id"]).sum(), work.duplicated(subset=["object_id"]).sum()],
    ["Έλλειψη ακριβούς αριθμητικού έτους", raw_year.isna().sum(), work["creation_year"].isna().sum()],
    ["Αριθμητικά έτη εκτός κανόνα", (raw_year.notna() & ((raw_year < 1870) | (raw_year > 1930) | (raw_year % 1 != 0))).sum(), (work["creation_year"].notna() & ((work["creation_year"] < 1870) | (work["creation_year"] > 1930))).sum()],
    ["Μη έγκυρα γνωστά πλήθη εικόνων", (raw_images.notna() & ((raw_images < 0) | (raw_images % 1 != 0))).sum(), (work["image_count"].notna() & ((work["image_count"] < 0) | (work["image_count"] % 1 != 0))).sum()],
    ["Ελλείψεις πλήθους εικόνων", raw_images.isna().sum(), work["image_count"].isna().sum()],
    ["Άγνωστος τόπος", raw_place.isna().sum(), work["place"].isna().sum()]
]
quality_report = pd.DataFrame(summary_rows, columns=["Μέτρηση", "Πριν (14 γραμμές)", "Μετά (11 γραμμές)"])
display(quality_report)
print("Ισοζύγιο γραμμών:", len(raw), "=", len(exact_removed), "+", len(conflicts), "+", len(work))
```

### Τι περιμένουμε να δούμε
- **14 → 11 γραμμές:** αφαιρέθηκε 1 ακριβές πλεονάζον αντίγραφο και εξαιρέθηκαν προσωρινά 2 αντικρουόμενες γραμμές ενός τεκμηρίου.
- **12 → 11 διαφορετικοί κωδικοί:** το OBJ007 παραμένει σε εκκρεμότητα, όχι χαμένο χωρίς εξήγηση.
- **3 → 2 ελλείψεις ακριβούς αριθμητικού έτους:** το 19O2 αποκαταστάθηκε τεκμηριωμένα. Το άγνωστο έτος και το προσεγγιστικό έτος παραμένουν χωρίς ακριβή αριθμητική τιμή.
- **1 → 0 μη έγκυρα αριθμητικά έτη:** το 2098 διορθώθηκε με REF-B.
- **1 → 0 μη έγκυρα γνωστά πλήθη εικόνων, αλλά 0 → 1 έλλειψη πλήθους:** το −2 δεν αντικαταστάθηκε με επινοημένο αριθμό.

**Περισσότερες ορατές ελλείψεις δεν σημαίνουν αναγκαστικά χειρότερα δεδομένα.** Μπορεί να σημαίνουν ότι μια μη έγκυρη τιμή αναγνωρίστηκε και δηλώθηκε με ειλικρίνεια.
