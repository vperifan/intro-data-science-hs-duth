```python
import pandas as pd
from io import StringIO

csv_text = """id,year,place,words
E01,1880,Αθήνα,120
E02,1881,Σμύρνη,180
E03,1882,Αθήνα,90
E04,1883,Κωνσταντινούπολη,210
E05,1884,Σμύρνη,150
E06,1885,Αθήνα,130
E07,1886,Κωνσταντινούπολη,240
E08,1887,Σμύρνη,110
E09,1888,Αθήνα,170
E10,1889,Κωνσταντινούπολη,200
E11,1890,Σμύρνη,160
E12,1891,Αθήνα,140
"""

letters = pd.read_csv(StringIO(csv_text))
letters
```

**Εντοπίστε στον πίνακα:** την επιστολή `E04`, το έτος της, τον τόπο αποστολής και τον αριθμό λέξεων. Οι αριθμοί στην αριστερή πλευρά είναι δείκτες της pandas και δεν αποτελούν πεδίο του CSV.
