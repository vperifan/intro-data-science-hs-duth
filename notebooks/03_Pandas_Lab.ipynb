## Προαιρετική επέκταση · Ένθετο JSON
Μια εγγραφή μπορεί να έχει λίστα θεμάτων. Η pandas μπορεί να κρατήσει τη λίστα μέσα σε ένα κελί, χωρίς να δημιουργήσει αυτόματα μία γραμμή ανά θέμα.

Στο παράδειγμα υπάρχουν **δύο τεκμήρια** αλλά συνολικά **τρεις καταχωρίσεις θεμάτων**. Αν αλλάζαμε τη δομή σε μία γραμμή ανά θέμα, θα άλλαζε η μονάδα του πίνακα. Η διάκριση είναι ίδια με εκείνη ανάμεσα σε τεκμήρια και εικόνες της θεωρίας.

````python
nested_records = [
    {"object_id": "OBJ001", "subjects": ["λιμάνι", "πλοία"]},
    {"object_id": "OBJ003", "subjects": ["ακτή"]}
]
nested_df = pd.DataFrame(nested_records)
display(nested_df)
print("Εγγραφές τεκμηρίων:", len(nested_df))
````
