## 1. Μεταβλητές και τύποι
Μια μεταβλητή είναι ένα όνομα που χρησιμοποιούμε για να αναφερθούμε σε μια τιμή. Το `=` αναθέτει την τιμή στα δεξιά στο όνομα στα αριστερά.

Στο παράδειγμα, κάθε τιμή περιγράφει μια **υποθετική επιστολή**. Η `print()` εμφανίζει τιμές και η `type()` δείχνει τον τύπο τους.

Πριν εκτελέσεις: ποια μεταβλητή είναι αριθμός και ποια κείμενο;

```python
year = 1880
place = "Αθήνα"
words = 120
is_digitised = True

print(place)
print(words)
print(type(year))
print(type(place))
print(type(is_digitised))
```

**Αναμενόμενα:** `Αθήνα`, `120`, και οι τύποι `int`, `str`, `bool`.

| Τύπος | Παράδειγμα | Χρήση |
|---|---|---|
| `int` | `120` | Ακέραιος |
| `float` | `12.5` | Αριθμός κινητής υποδιαστολής |
| `str` | `"Αθήνα"` | Κείμενο |
| `bool` | `True` | Λογική τιμή |

Στην Python οι δεκαδικοί γράφονται με **τελεία**. Τα εισαγωγικά δηλώνουν κείμενο. Τα κεφαλαία και τα πεζά στα ονόματα διαφέρουν.

```python
print(120 + 30)
print("120" + "30")

words_text = "120"
converted_words = int(words_text)
print(converted_words + 30)
```

**Αναμενόμενη έξοδος:** `150`, `12030`, `150` σε τρεις γραμμές.

Η δεύτερη πράξη ενώνει κείμενα. Η `int("120")` μετατρέπει κατάλληλο κείμενο σε ακέραιο. Η μετατροπή `int("άγνωστο")` θα προκαλούσε σφάλμα· δεν την εκτελούμε εδώ.

## 2. Λίστες και υπολογισμοί
Η λίστα κρατά πολλές τιμές με συγκεκριμένη σειρά. Τα στοιχεία γράφονται σε αγκύλες και χωρίζονται με κόμματα.

Οι επόμενες έξι τιμές είναι οι αριθμοί λέξεων των πρώτων έξι συνθετικών επιστολών της 1ης εβδομάδας. **Μία τιμή αντιστοιχεί σε μία επιστολή.**

```python
word_counts = [120, 180, 90, 210, 150, 130]

print("Λίστα:", word_counts)
print("Πλήθος επιστολών:", len(word_counts))
print("Πρώτη τιμή:", word_counts[0])
print("Τρίτη τιμή:", word_counts[2])
print("Τελευταία τιμή:", word_counts[-1])
print("Τμήμα θέσεων 1 έως 3:", word_counts[1:4])
```

**Έλεγξε:** πλήθος `6`, πρώτη τιμή `120`, τρίτη `90`, τελευταία `130`, τμήμα `[180, 90, 210]`.

Η πρώτη θέση έχει δείκτη **0**. Στο `[1:4]`, η θέση 4 **δεν περιλαμβάνεται**. Η επιλογή τμήματος δεν διαγράφει στοιχεία από το αρχικό `word_counts`.

Η `len()` μετρά στοιχεία, ενώ η `sum()` προσθέτει τις τιμές τους.

```python
total_words = sum(word_counts)
number_of_letters = len(word_counts)
mean_words = total_words / number_of_letters

print("Σύνολο λέξεων:", total_words)
print("Μέση τιμή:", round(mean_words, 2))
```

**Αναμενόμενα:** `880` λέξεις συνολικά και περίπου `146.67` λέξεις ανά επιστολή. Ο υπολογισμός της μέσης τιμής εδώ προϋποθέτει ότι η λίστα δεν είναι κενή.

### Αλλαγή σε αντίγραφο
Η `append()` προσθέτει στοιχείο στο τέλος. Για να μη μεταβάλουμε την αρχική λίστα αριθμών, δημιουργούμε αντίγραφο με `.copy()`.

Το παράδειγμα αναδημιουργεί το αντίγραφο σε κάθε εκτέλεση, ώστε να μην προσθέτει επανειλημμένα την τιμή 240.

```python
working_counts = word_counts.copy()
working_counts.append(240)

print("Αρχική:", word_counts)
print("Αντίγραφο:", working_counts)
print("Πλήθος αντιγράφου:", len(working_counts))
```

**Αναμενόμενο:** η αρχική λίστα έχει 6 τιμές και το αντίγραφο 7. Το `working_counts = word_counts` θα έδινε δύο ονόματα στην ίδια λίστα, χωρίς ανεξάρτητο αντίγραφο.

Η `.copy()` αρκεί για αυτό το παράδειγμα με αριθμούς. Δεν αποτελεί βαθύ αντίγραφο εμφωλευμένων δομών.

## 3. Συνθήκες
Μια σύγκριση δίνει `True` ή `False`. Το `if` χρησιμοποιεί αυτό το αποτέλεσμα για να επιλέξει εντολές.

| Έκφραση | Σημασία |
|---|---|
| `words > 150` | Περισσότερες από 150 λέξεις |
| `words >= 150` | Τουλάχιστον 150 λέξεις |
| `words == 150` | Ακριβώς 150 λέξεις |

Το `=` κάνει ανάθεση, ενώ το `==` σύγκριση. Οι εσοχές και η `:` είναι μέρος της σύνταξης.

```python
example_words = 150

if example_words >= 150:
    print("Τουλάχιστον 150 λέξεις")
else:
    print("Κάτω από 150 λέξεις")
```

## 4. Επαναλήψεις
Το `for` παίρνει διαδοχικά κάθε στοιχείο της λίστας. Οι εντολές με εσοχή εκτελούνται για κάθε στοιχείο.

```python
word_counts = [120, 180, 90, 210, 150, 130]

for current_words in word_counts:
    print(current_words)
```

Η έξοδος έχει έξι γραμμές, μία για κάθε τιμή. Το `current_words` παίρνει πρώτα το 120, μετά το 180 και συνεχίζει μέχρι το 130.

### Παράδειγμα · Πόσες επιστολές έχουν τουλάχιστον 150 λέξεις;
Μηδενίζουμε τον μετρητή **πριν** από το `for`. Για κάθε τιμή που ικανοποιεί τη συνθήκη, τον αυξάνουμε κατά ένα.

```python
count_at_least_150 = 0

for current_words in word_counts:
    if current_words >= 150:
        count_at_least_150 = count_at_least_150 + 1

print("Επιστολές με τουλάχιστον 150 λέξεις:", count_at_least_150)
```

**Αναμενόμενο:** `3`, για τις τιμές 180, 210 και 150.

Αν θέλουμε τις ίδιες τις επιλεγμένες τιμές, δημιουργούμε νέα λίστα αντί για έναν μόνο μετρητή.

```python
selected_word_counts = []

for current_words in word_counts:
    if current_words >= 150:
        selected_word_counts.append(current_words)

print(selected_word_counts)
```

**Αναμενόμενο:** `[180, 210, 150]`.

## Συνάρτηση με παράμετρο
Η ενότητα αυτή είναι για πρόσθετη εξάσκηση.

Μπορούμε να οργανώσουμε την καταμέτρηση σε συνάρτηση. Οι παράμετροι `values` και `limit` δέχονται τη λίστα και το όριο. Το `return` επιστρέφει το πλήθος.

Το παράδειγμα χρησιμοποιεί δική του λίστα και δεν εξαρτάται από τις απαντήσεις των ασκήσεων.

```python
def count_at_least(values, limit):
    count = 0
    for value in values:
        if value >= limit:
            count = count + 1
    return count

extension_minutes = [40, 25, 60, 30, 15, 50]
print("Τουλάχιστον 30:", count_at_least(extension_minutes, 30))
print("Τουλάχιστον 45:", count_at_least(extension_minutes, 45))
print("Κενή λίστα:", count_at_least([], 30))
```
